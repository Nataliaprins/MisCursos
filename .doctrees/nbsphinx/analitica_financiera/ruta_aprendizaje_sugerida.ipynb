# Apéndice: Ruta de Aprendizaje Sugerida

Este curso cubre los fundamentos de la analítica financiera aplicada con Python: ingeniería de datos financieros, modelos predictivos y redes neuronales para series de tiempo. Para quienes quieran **profundizar y especializarse** en el uso de Python en temas financieros, esta es una ruta de cursos externos (la mayoría gratuitos o auditables sin costo) organizada de forma progresiva, desde fundamentos de programación hasta *deep learning* aplicado a series de tiempo financieras.

No es necesario completar toda la ruta ni seguirla en un orden estricto: elige el punto de entrada según tu nivel actual y tus intereses.

## F0 - Fundamentos de Python y manejo de datos

Punto de partida si aún no tienes experiencia programando o quieres repasar lo esencial de Python y `pandas` antes de continuar con contenido financiero.

- **[Python](https://www.kaggle.com/learn/python)** (Kaggle Learn) — sintaxis básica, funciones, estructuras de control.
- **[Pandas](https://www.kaggle.com/learn/pandas)** (Kaggle Learn) — manipulación de `DataFrames`, la herramienta que usamos en todo el curso.
- **[Python for Everybody](https://www.coursera.org/specializations/python)** (Coursera, U. de Michigan) — especialización completa; para esta ruta basta con los dos primeros cursos:
  - [Programming for Everybody (Getting Started with Python)](https://www.coursera.org/learn/python)
  - [Python Data Structures](https://www.coursera.org/learn/python-data)

## F1 - De datos crudos a variables con valor predictivo

Profundiza en la construcción de variables (*feature engineering*) y en su aplicación específica a la construcción de portafolios de inversión.

- **[Feature Engineering](https://www.kaggle.com/learn/feature-engineering)** (Kaggle Learn) — técnicas generales de creación de variables.
- **[Introduction to Portfolio Construction and Analysis with Python](https://www.coursera.org/learn/introduction-portfolio-construction-python)** (Coursera, EDHEC Business School) — teoría moderna de portafolios (Markowitz, CAPM) implementada en Python; conecta directamente con el módulo de *Portafolios de Inversión* de este curso.

## F2 - Machine learning aplicado a inversiones

Una vez cómodo con Python y los fundamentos de datos financieros, esta fase introduce *machine learning* de propósito general y su aplicación específica a la gestión de activos.

- **[Machine Learning Specialization](https://www.coursera.org/specializations/machine-learning-introduction)** (Coursera, DeepLearning.AI / Stanford, Andrew Ng) — para esta ruta, con los cursos 2 y 3 es suficiente:
  - [Advanced Learning Algorithms](https://www.coursera.org/learn/advanced-learning-algorithms) (redes neuronales, árboles de decisión)
  - [Unsupervised Learning, Recommenders, Reinforcement Learning](https://www.coursera.org/learn/unsupervised-learning-recommenders-reinforcement-learning)
- **[Python and Machine Learning for Asset Management](https://www.coursera.org/learn/python-machine-learning-for-investment-management)** (Coursera, EDHEC Business School) — aplica esos algoritmos directamente a la gestión de portafolios.

## F3 - Inteligencia artificial generativa y lenguaje natural

Fase orientada a quienes quieran incorporar IA generativa (LLMs) al análisis de información financiera no estructurada (noticias, reportes, transcripciones de resultados).

- **[Generative AI with Large Language Models](https://www.coursera.org/learn/generative-ai-with-llms)** (Coursera, DeepLearning.AI / AWS) — cómo se entrenan, ajustan y despliegan los modelos de lenguaje actuales.
- **[Natural Language Processing Specialization](https://www.coursera.org/specializations/natural-language-processing)** (Coursera, DeepLearning.AI) — *opcional*, para quienes quieran profundizar en NLP clásico y moderno antes o después del curso de LLMs.

## F4 - Deep learning para series de tiempo

Cierra el círculo con el tema del cuaderno anterior de este curso (*Redes Neuronales y Deep Learning para Series de Tiempo Financieras*), profundizando en arquitecturas y buenas prácticas específicas para pronóstico de series temporales.

- **[Intro to Deep Learning](https://www.kaggle.com/learn/intro-to-deep-learning)** (Kaggle Learn) — fundamentos de redes neuronales con TensorFlow/Keras.
- **[Time Series](https://www.kaggle.com/learn/time-series)** (Kaggle Learn) — técnicas de pronóstico, más allá del enfoque de redes neuronales.
- **[Sequences, Time Series and Prediction](https://www.coursera.org/learn/tensorflow-sequences-time-series-and-prediction)** (Coursera, DeepLearning.AI, curso 4 de la especialización TensorFlow Developer) — LSTM, GRU y capas convolucionales aplicadas a series de tiempo.

## Notas finales

- La mayoría de estos cursos pueden **auditarse gratis** en Coursera (acceso al contenido sin certificado); Kaggle Learn es siempre gratuito y no requiere inscripción previa.
- Las duraciones son aproximadas y corresponden al ritmo sugerido por cada plataforma; con dedicación constante, la ruta completa (F0 a F4) es abordable en un semestre.
- Si solo te interesa el tema de este curso en particular, la combinación mínima recomendada es **F0 → F1 → F4**: fundamentos, portafolios y series de tiempo con redes neuronales.
