## Seccion A. Seleccion multiple (20 preguntas)

### Preguntas 1 a 10 (Teoricas con caso aplicado)

1. **Caso:** Una cadena de supermercados quiere explicar las ventas mensuales segun precio promedio, inversion en publicidad y trafico de clientes. La gerencia pide un modelo interpretable para apoyar decisiones.
   - A) K-means
   - B) Regresion lineal multiple
   - C) Arbol de clasificacion
   - D) Analisis factorial

2. **Caso:** Una empresa de seguros modela siniestros esperados por cliente y sospecha que el efecto de la edad cambia segun el tipo de plan.
   - A) Eliminar la variable edad
   - B) Incluir un termino de interaccion edad x tipo de plan
   - C) Usar solo correlaciones
   - D) Estandarizar y detener el analisis

3. **Caso:** Un banco estima mora crediticia (en porcentaje) y observa que la varianza de los errores aumenta para clientes de mayor ingreso.
   - A) No hacer nada, siempre es normal
   - B) Revisar heterocedasticidad y considerar errores robustos o transformaciones
   - C) Eliminar ingresos del modelo
   - D) Cambiar la variable dependiente por una cualitativa sin justificacion

4. **Caso:** Una empresa de retail desea saber si una promocion tuvo impacto en ventas controlando por estacionalidad y precio.
   - A) Regresion con variable dummy de promocion
   - B) Regresion sin intercepto ni control
   - C) Solo grafico de barras
   - D) Promedio simple antes y despues sin controles

5. **Caso:** En una empresa B2B, el director comercial pregunta por que el coeficiente de publicidad no fue significativo (p > 0.05).
   - A) Significa que publicidad no tiene ningun efecto en todos los contextos
   - B) Significa que con esta muestra y este modelo no hay evidencia estadistica suficiente
   - C) Significa que el modelo esta perfecto
   - D) Significa que el R2 siempre debe ser 1

6. **Caso:** Una firma de logistica estima costos operativos y reporta R2 muy alto, pero el equipo detecta residuos con patrones.
   - A) Un R2 alto garantiza validez completa
   - B) Deben revisarse supuestos y diagnosticos de residuos
   - C) Se debe eliminar el intercepto
   - D) El analisis termina con R2

7. **Caso:** Una empresa exportadora incluye tipo de cambio y precio internacional del commodity en el mismo modelo; la correlacion entre ambas es muy alta.
   - A) Posible multicolinealidad que afecta precision de coeficientes
   - B) Error de digitacion unicamente
   - C) Evidencia de causalidad definitiva
   - D) Prueba de normalidad directa

8. **Caso:** Una app de delivery quiere modelar tiempo de entrega y sospecha observaciones atipicas por dias de lluvia extrema.
   - A) Ignorar outliers siempre
   - B) Evaluar leverage, distancia de Cook e impacto en coeficientes
   - C) Reemplazar todos los datos extremos por el promedio
   - D) Eliminar aleatoriamente 20% de la muestra

9. **Caso:** Una empresa de manufactura necesita predecir demanda para planificar compras del siguiente trimestre.
   - A) Priorizar solo interpretacion y no evaluar error predictivo
   - B) Evaluar metricas como MAE o RMSE en datos de validacion
   - C) Usar cualquier modelo sin separar entrenamiento y prueba
   - D) Usar el modelo con menor numero de variables siempre

10. **Caso:** Una compania mide satisfaccion (1 a 10) y desea explicar su variacion por tiempo de espera, calidad percibida y canal.
   - A) Regresion lineal con variables numericas y dummies para canal
   - B) Solo tabla dinamica
   - C) Prueba t para cada variable por separado y sin modelo
   - D) Arbol de decision sin validar

### Preguntas 11 a 20 (Practicas)

11. En un modelo: $\hat{Y} = 120 - 3X_1 + 0.8X_2$, si $X_1$ aumenta en 1 unidad y $X_2$ permanece constante, la prediccion de $Y$:
   - A) Aumenta en 3
   - B) Disminuye en 3
   - C) No cambia
   - D) Aumenta en 0.8

12. En una regresion, el intervalo de confianza al 95% para $\beta_1$ es [0.2, 1.1]. Esto implica que:
   - A) $\beta_1$ es no significativo al 5%
   - B) $\beta_1$ es significativo al 5%
   - C) El modelo no sirve
   - D) El error estandar es cero

13. Se estimo un modelo con 4 predictores y 100 observaciones. Si se agregan variables irrelevantes, es mas probable que:
   - A) El R2 ajustado baje o no mejore
   - B) El R2 siempre baje
   - C) El R2 ajustado siempre suba
   - D) El error residual desaparezca

14. Una empresa obtiene VIF de 12 para una variable explicativa. La interpretacion mas adecuada es:
   - A) No hay problema
   - B) Posible multicolinealidad alta
   - C) Error de heterocedasticidad
   - D) Normalidad perfecta

15. Si en un test de Breusch-Pagan se obtiene p-valor = 0.01, al 5% se concluye:
   - A) No hay heterocedasticidad
   - B) Hay evidencia de heterocedasticidad
   - C) Hay autocorrelacion negativa
   - D) El modelo no tiene intercepto

16. En un modelo de ventas, la variable dummy "campana_digital" toma 1 si hubo campana y 0 si no. Su coeficiente estimado fue 18.
   - A) Con campana, ventas esperadas suben 18 unidades, ceteris paribus
   - B) Ventas bajan 18 unidades con campana
   - C) La dummy no se interpreta
   - D) El intercepto desaparece

17. Si el estadistico Durbin-Watson es cercano a 2, normalmente sugiere:
   - A) Fuerte autocorrelacion positiva
   - B) Ausencia aproximada de autocorrelacion de primer orden
   - C) Multicolinealidad severa
   - D) Sobreajuste garantizado

18. Dos modelos de pronostico se evaluan en test: Modelo A (RMSE=9.5), Modelo B (RMSE=7.8). Para prediccion puntual:
   - A) Conviene A por ser mas simple
   - B) Conviene B por menor error
   - C) Son equivalentes
   - D) No se puede comparar RMSE

19. En una regresion log-lineal: $\ln(Y)=\beta_0+\beta_1 X$, con $\beta_1=0.04$, un aumento de 1 unidad en $X$ implica aproximadamente:
   - A) Disminucion de 4% en $Y$
   - B) Aumento de 4% en $Y$
   - C) Aumento de 0.04 unidades en $Y$
   - D) Ningun cambio en $Y$

20. Si al revisar residuos vs ajustados se observa forma de abanico, la accion inicial mas apropiada es:
   - A) Confirmar posible heterocedasticidad y probar correcciones
   - B) Aceptar modelo sin cambios
   - C) Borrar la mitad de los datos
   - D) Cambiar a clasificacion automaticamente

## Seccion B. Falso / Verdadero (10 preguntas)

21. En regresion lineal, un coeficiente significativo estadisticamente prueba causalidad por si solo.

22. El supuesto de no multicolinealidad perfecta implica que ninguna variable explicativa puede escribirse como combinacion lineal exacta de otras.

23. Si el p-valor del test F global es menor a 0.05, al menos un coeficiente distinto del intercepto aporta explicacion en el modelo.

24. Un R2 alto siempre garantiza buena capacidad de pronostico fuera de muestra.

25. La presencia de heterocedasticidad puede volver ineficientes los errores estandar clasicos y afectar inferencia.

26. En una regresion multiple, el coeficiente de una variable se interpreta manteniendo constantes las demas variables del modelo.

27. El uso de variables dummy permite incorporar factores cualitativos, como region o tipo de cliente, en un modelo lineal.

28. El sobreajuste ocurre cuando un modelo capta ruido de entrenamiento y puede rendir peor en datos nuevos.

29. Si los residuos muestran patron sistematico frente a los valores ajustados, puede haber especificacion incorrecta del modelo.

30. En evaluacion predictiva, comparar MAE o RMSE en un conjunto de prueba ayuda a seleccionar mejor modelo para pronostico.

   - C) Prueba chi-cuadrado
   - D) Metodo Delphi

   - B) Eliminar variable ciudad
   - C) Usar solo media historica

   - C) Forzar R2 a subir

   - C) Eliminar datos de promociones
   - D) Analizar solo una semana al azar

   - C) La variable debe duplicarse

   - D) Ajustar modelo sin variable dependiente

   - D) Solo analisis cualitativo
