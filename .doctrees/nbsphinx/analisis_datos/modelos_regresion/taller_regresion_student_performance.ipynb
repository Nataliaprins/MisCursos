# Taller: Regresión Lineal con Student Performance

Este taller está diseñado para estudiantes de pregrado y cubre:

1. Regresión lineal simple.
2. Regresión lineal múltiple.
3. Comprobación de supuestos de los modelos.

Base de datos: [Rendimiento estudiantil](Student_Performance.csv)


**Caso de estudio:** Analítica académica para mejorar el desempeño estudiantil

Contexto pedagógico:

1.	**Consideraciones para el desarrollo del taller**

Este taller busca que los estudiantes apliquen los conceptos vistos en clase en un contexto cercano al trabajo real de un modelador de datos. Es posible que algunas preguntas involucren temas que no han sido abordados directamente en el curso; en esos casos, se espera que el estudiante investigue de manera autónoma utilizando fuentes confiables. El propósito es ampliar el espectro temático del curso, fortalecer las habilidades de aprendizaje independiente y desarrollar la capacidad de enfrentar problemas reales donde es necesario complementar continuamente los conocimientos adquiridos en clase.
Tome el conjunto de datos Student Performance y realice la construcción, interpretación y validación de modelos predictivos con regresión lineal simple y múltiple, basándose en las preguntas direccionadoras.


2.	**Contexto del caso**

La Universidad Horizonte enfrenta una preocupación creciente: el rendimiento académico promedio del primer año ha mostrado alta variabilidad entre cohortes y facultades. La Vicerrectoría Académica ha creado un equipo de analítica educativa para identificar qué factores están asociados con el desempeño final de los estudiantes y diseñar intervenciones tempranas.
Para iniciar el proyecto, el equipo dispone de un conjunto de datos históricos que incluye variables relacionadas con:

* Horas de estudio semanales.
* Asistencia a clases.
* Promedios o calificaciones previas.
* Participación en actividades académicas.
* Acceso a tutorías.
* Características demográficas básicas (cuando existan en el conjunto de datos).
* La variable objetivo: calificación final o indicador de desempeño académico.

Usted forma parte del equipo de analítica educativa y debe desarrollar un modelo de regresión lineal que ayude a explicar y predecir el desempeño académico final.



## Preguntas direccionadoras para el desarrollo del ejercicio

A. Comprensión del problema y de los datos

1.	¿Cuál es exactamente la variable objetivo que se desea explicar o predecir?
2.	¿Qué significado práctico tiene una diferencia de 1 punto en la variable objetivo?
3.	¿Qué variables podrían tener una relación asociativa con el desempeño académico?
4.	¿Existen variables que puedan actuar como confusoras (por ejemplo, rendimiento previo y horas de estudio)?
5.	¿Qué unidades tienen las variables numéricas (horas, porcentaje de asistencia, puntajes, etc.)?

B. Calidad y preparación de los datos
1.	¿Hay valores faltantes? ¿Cómo deberían tratarse?
2.	¿Existen observaciones atípicas o errores de captura?
3.	¿Es necesario transformar alguna variable (logaritmos, estandarización, escalamiento)?
4.	¿Cómo se codificarán las variables categóricas si están presentes?
5.	¿La distribución de la variable objetivo sugiere problemas importantes para una regresión lineal?

C. Exploración inicial
1.	¿Qué variables muestran mayor correlación con el desempeño final?
2.	¿Las relaciones observadas parecen lineales o podrían ser no lineales?
3.	¿Hay evidencia visual de heterocedasticidad o de segmentos claramente distintos de estudiantes?
4.	¿Qué variables parecen redundantes entre sí?

D. Construcción del modelo
1.	¿Cuál será el modelo base (por ejemplo, una sola variable explicativa)? Implemente un modelo de regresión lineal simple.
2.	¿Qué criterio se usará para incorporar variables adicionales: teoría, significancia estadística, desempeño predictivo o combinación de criterios? Implemente un modelo de regresión lineal multiple.
3.	¿Existen problemas de multicolinealidad? ¿Cómo se detectarán?

E. Evaluación del modelo
1.	¿Qué métricas se utilizarán para evaluar el ajuste (R², R² ajustado)?
2.	¿Qué indican los residuos sobre los supuestos de la regresión lineal?
3.	¿Hay observaciones influyentes que alteren de manera importante los coeficientes?

F.  Interpretación para la toma de decisiones
1.	¿Cómo se interpreta cada coeficiente relevante de las regresiones en lenguaje no técnico?
2.	¿Qué variables muestran la asociación más fuerte con el desempeño final?
3.	¿Qué resultados son estadísticamente significativos y cuáles no?


La solución del taller debe desarrollarse íntegramente en lenguaje python. Como evidencia del trabajo realizado,  deberá entregar el Notebook (puede compartir el cuaderno de colab) con el código completamente documentado y ejecutable, así como un informe técnico en PDF en el que se presenten, interpreten y discutan los resultados obtenidos. El informe debe incluir las conclusiones y recomendaciones derivadas del análisis realizado.


**El informe del taller debe incluir, como mínimo:**

1.	Descripción del problema y de la variable objetivo.
2.	Resumen de limpieza y preparación de datos.
3.	Análisis exploratorio (tablas y visualizaciones relevantes).
4.	Modelo(s) de regresión estimados.
5.	Tabla de coeficientes e interpretación contextualizada.
6.	Evaluación del desempeño predictivo en entrenamiento y prueba.
7.	Diagnóstico de residuos y discusión de supuestos.
8.	Conclusiones ejecutivas (3–5 hallazgos clave).
9.	Recomendaciones de intervención académica sustentadas por el análisis.

### Criterios transversales de descuento

- **Falta de interpretación contextual** (solo salida técnica sin análisis): hasta **-10 puntos**.  
- **Ausencia de justificación metodológica** en decisiones clave: hasta **-10 puntos**.
