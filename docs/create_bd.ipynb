```python
# ============================================================
# SIMULACIÓN DE BASE DE DATOS PARA DISEÑO DE EXPERIMENTOS
# Curso: Análisis de Datos para Administración de Empresas
# Temas:
# 1. Diseño Completamente Aleatorizado (DCA)
# 2. Diseño en Bloques Aleatorizados (DBA)
#
# Caso empresarial:
# Una empresa quiere evaluar el efecto de diferentes
# estrategias de marketing sobre las ventas.
# ============================================================

import numpy as np
import pandas as pd

# Reproducibilidad
np.random.seed(123)

# ============================================================
# 1. DISEÑO COMPLETAMENTE ALEATORIZADO (DCA)
# ============================================================

# Tratamientos:
# Diferentes estrategias de marketing
tratamientos = [
    "Email Marketing",
    "Publicidad Redes",
    "Influencers",
    "Descuentos"
]

# Número de observaciones por tratamiento
n = 30

datos_dca = []

# Efectos promedio simulados sobre ventas
efectos = {
    "Email Marketing": 200,
    "Publicidad Redes": 240,
    "Influencers": 280,
    "Descuentos": 260
}

for tratamiento in tratamientos:
    
    for i in range(n):
        
        # Variabilidad aleatoria
        error = np.random.normal(0, 25)
        
        ventas = efectos[tratamiento] + error
        
        datos_dca.append({
            "Tratamiento": tratamiento,
            "Ventas": round(ventas, 2)
        })

# Crear DataFrame
df_dca = pd.DataFrame(datos_dca)

print("===================================================")
print("BASE DE DATOS - DISEÑO COMPLETAMENTE ALEATORIZADO")
print("===================================================")

print(df_dca.head())

# Guardar CSV
df_dca.to_csv("diseno_completamente_aleatorizado.csv", index=False)



# ============================================================
# 2. DISEÑO DE BLOQUES ALEATORIZADOS (DBA)
# ============================================================

# Supuesto:
# Las ventas pueden depender también de la ciudad.
# Cada ciudad será un bloque.

bloques = [
    "Bogotá",
    "Medellín",
    "Cali",
    "Barranquilla"
]

# Efectos de bloque (ciudad)
efectos_bloque = {
    "Bogotá": 40,
    "Medellín": 20,
    "Cali": -10,
    "Barranquilla": -20
}

datos_dba = []

# Número de repeticiones por combinación
rep = 8

for bloque in bloques:
    
    for tratamiento in tratamientos:
        
        for i in range(rep):
            
            # Error aleatorio
            error = np.random.normal(0, 20)
            
            # Modelo:
            # media general + efecto tratamiento + efecto bloque + error
            
            ventas = (
                200
                + (efectos[tratamiento] - 200)
                + efectos_bloque[bloque]
                + error
            )
            
            datos_dba.append({
                "Ciudad_Bloque": bloque,
                "Tratamiento": tratamiento,
                "Ventas": round(ventas, 2)
            })

# Crear DataFrame
df_dba = pd.DataFrame(datos_dba)

print("\n===================================================")
print("BASE DE DATOS - DISEÑO DE BLOQUES ALEATORIZADOS")
print("===================================================")

print(df_dba.head())

# Guardar CSV
df_dba.to_csv("diseno_bloques_aleatorizados.csv", index=False)



# ============================================================
# RESUMEN DESCRIPTIVO
# ============================================================

print("\n===================================================")
print("RESUMEN DCA")
print("===================================================")

print(
    df_dca.groupby("Tratamiento")["Ventas"]
    .agg(["mean", "std", "count"])
)

print("\n===================================================")
print("RESUMEN DBA")
print("===================================================")

print(
    df_dba.groupby(["Ciudad_Bloque", "Tratamiento"])["Ventas"]
    .mean()
)
```
