## 1. Intervalos de confianza para la media



### 1.1 Caso con desviación estándar poblacional conocida



**Situación.** Una empresa desea estimar el gasto promedio en miles de pesos de clientes impactados por una campaña digital. Estudios históricos indican que la desviación estándar poblacional es de 12. En una muestra de 64 clientes se obtuvo un gasto promedio de 85.



**Tarea.** Construya un intervalo de confianza del 95% para la media poblacional e interprételo desde mercadeo.



La fórmula es:



$$

\bar{x} \pm z_{\alpha/2} \frac{\sigma}{\sqrt{n}}

$$


### 1.2 Caso con desviación estándar poblacional desconocida



**Situación.** Un equipo de trade marketing mide el tiempo promedio, en minutos, que un cliente permanece en un punto de venta intervenido. En una muestra de 25 clientes se obtiene una media de 18 minutos y una desviación estándar muestral de 4.5 minutos.



**Tarea.** Construya un intervalo de confianza del 95% usando la distribución t de Student.



La fórmula es:



$$

\bar{x} \pm t_{\alpha/2, n-1} \frac{s}{\sqrt{n}}

$$


### 1.3 Relación entre nivel de confianza, error estándar y amplitud del intervalo



**Idea clave.** La amplitud del intervalo aumenta cuando:



- aumenta el nivel de confianza,

- aumenta la variabilidad,

- disminuye el tamaño de muestra.

## 2. Intervalos de confianza vs intervalos de probabilidad



### Diferencias conceptuales



Un **intervalo de confianza** busca estimar un parámetro desconocido, por ejemplo la media poblacional de gasto.



Un **intervalo de probabilidad** o de predicción busca contener el valor de una observación futura individual con una probabilidad dada.



En mercadeo, esto implica una diferencia importante:



- el intervalo de confianza sirve para estimar el comportamiento promedio del mercado,

- el intervalo de predicción sirve para anticipar el rango posible de respuesta de un cliente individual.



**Ejemplo aplicado.** Se registra el monto de compra de 40 clientes expuestos a una campaña de email marketing. La muestra tiene media 110 y desviación estándar 20.


## 3. Intervalos de confianza para proporciones



### 3.1 Proporción muestral y construcción del intervalo en muestras grandes



**Situación.** De 500 usuarios impactados por una campaña en redes sociales, 290 hicieron clic en el anuncio. Se desea estimar la proporción poblacional de clientes que responderían positivamente a la campaña.



La proporción muestral es:



$$

\hat{p} = \frac{x}{n}

$$



Y el intervalo de confianza para muestras grandes es:



$$

\hat{p} \pm z_{\alpha/2}\sqrt{\frac{\hat{p}(1-\hat{p})}{n}}

$$


## 4. Determinación del tamaño de muestra



### 4.1 Para estimar la media poblacional



Si se conoce o se aproxima la desviación estándar, el tamaño de muestra puede calcularse como:



$$

n = \left(\frac{z_{\alpha/2}\sigma}{E}\right)^2

$$



donde $E$ es el error máximo permitido.



**Situación.** Una cadena de retail desea estimar el gasto promedio por visita con 95% de confianza, error máximo de 2 miles de pesos y desviación estándar histórica de 10.



### 4.2 Para estimar una proporción



En muestras grandes:



$$

n = \frac{z_{\alpha/2}^2 p(1-p)}{E^2}

$$



Si no se conoce $p$, se usa $p = 0.5$ para obtener el tamaño más conservador.



**Situación.** Una empresa quiere estimar la proporción de clientes que reconocerían una nueva marca con 95% de confianza y error máximo de 4%.



### Factores que influyen



- Mayor nivel de confianza implica mayor tamaño de muestra.

- Menor error máximo permitido implica mayor tamaño de muestra.

- Mayor variabilidad también exige mayor tamaño de muestra.
