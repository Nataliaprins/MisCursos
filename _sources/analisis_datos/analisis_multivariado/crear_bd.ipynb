```python
import pandas as pd
import numpy as np

# Cargar base original
# Asegúrate de tener el archivo "base_kpis_empresa.csv" en el mismo directorio

df = pd.read_csv("base_kpis_empresa.csv")
df_sucia = df.copy()

# Semilla para reproducibilidad
rng = np.random.default_rng(42)

# Seleccionar aleatoriamente 10 columnas (KPI) para insertar valores faltantes
num_kpis = min(10, len(df_sucia.columns))
columnas_kpi = rng.choice(df_sucia.columns, size=num_kpis, replace=False)

for col in columnas_kpi:
    # Elegir aleatoriamente entre 1 y 5 valores faltantes por columna
    n_missing = rng.integers(1, 6)
    idx_missing = rng.choice(df_sucia.index, size=n_missing, replace=False)
    df_sucia.loc[idx_missing, col] = np.nan

nombres_kpis = [
    "OP_Ciclo_Produccion", "OP_Tasa_Defectos", "OP_Disponibilidad_Maq", "OP_Costo_Unitario", 
    "OP_Merma_Materia_Prima", "OP_Cumplimiento_Plan", "OP_Eficiencia_Energetica", "OP_Rotacion_Inventario", 
    "OP_Tiempo_Set_Up", "OP_Productividad_HH", "OP_Lead_Time_Prov", "OP_Capacidad_Utilizada",
    "FIN_Margen_EBITDA", "FIN_ROI", "FIN_Flujo_Caja_Libre", "FIN_Apalancamiento", 
    "FIN_Dias_Cuentas_Cobrar", "FIN_Dias_Pago_Prov", "FIN_Costo_Capital_WACC", "FIN_Margen_Neto", 
    "FIN_Ratio_Liquidez", "FIN_Variacion_Gastos_Adm", "FIN_Punto_Equilibrio", "FIN_EBIT", "FIN_Costo_Deuda",
    "MKT_NPS", "MKT_CAC", "MKT_LTV", "MKT_Tasa_Churn", "MKT_Conversion_Lead", "MKT_Market_Share", 
    "MKT_Ticket_Promedio", "MKT_Reclamaciones_QRS", "MKT_Reconocimiento_Marca", "MKT_Costo_Por_Click", 
    "MKT_Frecuencia_Compra", "MKT_Tasa_Recomendacion",
    "RH_Indice_Clima", "RH_Tasa_Rotacion", "RH_Ausentismo", "RH_Horas_Capacitacion", 
    "RH_Productividad_Empleado", "RH_Satisfaccion_Beneficios", "RH_Tiempo_Contratacion", 
    "RH_Gasto_Nomina_Ventas", "RH_Indice_Diversidad", "RH_Promociones_Internas", 
    "RH_Evaluacion_Desempeno", "RH_Accidentalidad_Laboral", "RH_Antiguedad_Promedio"
]

df_sucia.columns = nombres_kpis

# Guardar nueva base "sucia"
df_sucia.to_csv("base_kpis_sucia.csv", index=False)

# Resumen rápido
print("Archivo generado: base_kpis_sucia.csv")
print(f"Filas: {df_sucia.shape[0]}, Columnas: {df_sucia.shape[1]}")
print("Nulos por columna:")
print(df_sucia.isna().sum())
```
