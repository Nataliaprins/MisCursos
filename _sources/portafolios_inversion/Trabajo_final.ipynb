## Parte 4: Optimización de Markowitz (Media-Varianza)

### 4.1 Teoría de Markowitz

La optimización de **Media-Varianza** busca encontrar la combinación de activos que:
- **Maximiza el rendimiento** para un nivel de riesgo dado, o
- **Minimiza el riesgo** para un nivel de rendimiento dado

#### Problema de optimización:

$$\min_w \quad w^\top \Sigma w$$ sujeto a: $$w^\top \mu = R_{objetivo}$$ $$w^\top \mathbf{1} = 1$$ $$w_i \geq 0 \quad \forall i$$ 


**Donde:**

$w$ = vector de pesos

$\Sigma$ = matriz de covarianza

$\mu$ = vector de rendimientos esperados

$R_{objetivo}$ = rendimiento objetivo

#### Portafolios especiales:
1. **Portafolio de Mínima Varianza (GMV)**: Menor riesgo posible
2. **Portafolio Tangente (Máximo Sharpe)**: Mayor ratio Sharpe

## Parte 5: Modelo de Black-Litterman

### 5.1 Teoría de Black-Litterman

El modelo de **Black-Litterman (1992)** mejora la optimización de Markowitz al:

1. **Partir del equilibrio de mercado**: Usa las ponderaciones de capitalización como punto de partida neutral
2. **Incorporar views del inversionista**: Permite ajustar las expectativas de rendimiento de forma bayesiana

#### Componentes del modelo:

| Componente | Símbolo | Descripción |
|:---|:---:|:---|
| Rendimientos de equilibrio | $\Pi$ | $\Pi = \delta \Sigma w_{mkt}$ |
| Escalar de incertidumbre | $\tau$ | Típicamente entre 0.01 y 0.05 |
| Matriz de views | $P$ | Identifica activos en cada opinión |
| Vector de views | $Q$ | Rendimientos esperados por el inversionista |
| Incertidumbre de views | $\Omega$ | Confianza en cada opinión |

#### Fórmula de rendimientos combinados:

$$E(R) = \left[(\tau \Sigma)^{-1} + P^\top \Omega^{-1} P\right]^{-1} \left[(\tau \Sigma)^{-1} \Pi + P^\top \Omega^{-1} Q\right]$$

## Parte 6: Hierarchical Risk Parity (HRP)

### 6.1 Teoría de HRP

El algoritmo **Hierarchical Risk Parity (HRP)** de Marcos López de Prado (2016) evita la inversión de la matriz de covarianza mediante tres pasos:

#### Paso 1: Clustering Jerárquico
- Transforma correlaciones en distancias: $d_{i,j} = \sqrt{\frac{1}{2}(1 - \rho_{i,j})}$
- Agrupa activos similares usando linkage jerárquico

#### Paso 2: Quasi-Diagonalización
- Reordena la matriz de covarianza según la estructura del dendrograma
- Los activos similares quedan adyacentes

#### Paso 3: Bisección Recursiva
- Asigna pesos inversamente proporcionales a la varianza de cada cluster
- Divide recursivamente los clusters hasta llegar a activos individuales

#### Ventajas sobre Markowitz:
- **No requiere invertir** la matriz de covarianza
- **Más estable** ante errores de estimación
- **Diversificación intuitiva** basada en estructura de correlaciones

## Parte 7: Evaluación de Riesgo - Value at Risk (VaR)

### 7.1 Concepto de Value at Risk

El **Value at Risk (VaR)** responde a la pregunta:

> *"¿Cuál es la máxima pérdida esperada en un período dado, con un nivel de confianza determinado?"*

Por ejemplo, un **VaR(95%) = 2%** significa que existe un 95% de probabilidad de que la pérdida diaria no supere el 2%.

### Métodos de cálculo:

| Método | Descripción | Ventajas | Desventajas |
|:---|:---|:---|:---|
| **Paramétrico** | Asume distribución normal | Simple, rápido | Subestima colas |
| **Histórico** | Usa percentiles históricos | No asume distribución | Requiere datos |
| **Monte Carlo** | Simula escenarios | Flexible, captura no-linealidades | Computacionalmente costoso |

#### Fórmulas:

**VaR Paramétrico:**
$$VaR_{p\%} = \mu - z_{p\%} \cdot \sigma$$

**VaR Histórico:**
$$VaR_{p\%} = \text{Percentil}_{(1-p)\%} \text{ de rendimientos históricos}$$

---

### 7.2 Expected Shortfall (ES) o Conditional Value at Risk (CVaR)

El **Expected Shortfall (ES)**, también conocido como **Conditional Value at Risk (CVaR)**, es una medida de riesgo que responde a la pregunta:

> *"Si ocurre una pérdida mayor al VaR, ¿cuál es la pérdida promedio esperada?"*

A diferencia del VaR, el ES captura el **riesgo de cola** y es considerada una medida de riesgo **coherente** que satisface las propiedades de:
- Monotonicidad
- Subaditividad
- Homogeneidad positiva
- Invarianza ante traslaciones

#### Fórmulas:

**ES Paramétrico (asumiendo distribución normal):**
$$ES_{p\%} = \mu + \sigma \cdot \frac{\phi(z_{p\%})}{1-p}$$

Donde $\phi(z)$ es la función de densidad de probabilidad normal estándar.

**ES Histórico:**
$$ES_{p\%} = \frac{1}{n_{tail}} \sum_{r_i < VaR_{p\%}} r_i$$

Es decir, el promedio de todos los rendimientos que están por debajo del VaR.

#### Comparación VaR vs ES:

| Característica | VaR | ES (CVaR) |
|:---|:---|:---|
| **Interpretación** | Umbral de pérdida máxima | Pérdida promedio en cola |
| **Riesgo de cola** | No captura | Captura completamente |
| **Coherencia** | No es coherente | Es coherente |
| **Regulación** | Basilea II | Basilea III (preferido) |
| **Optimización** | No convexa | Convexa |
